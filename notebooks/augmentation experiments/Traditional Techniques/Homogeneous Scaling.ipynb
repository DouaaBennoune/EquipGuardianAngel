```python
fd001_cols=fd001.columns
fd002_cols=fd002.columns
fd003_cols=fd003.columns
fd004_cols=fd004.columns
```

```python
print(fd001_arr)
```

```python

def homogeneous_scaling(X, alpha_mean=1.0, alpha_std=0.1):
    """
    X : np.ndarray of shape (seq_len, n_features)
    returns scaled version
    """
    alpha = np.random.normal(loc=alpha_mean, scale=alpha_std)
    return X * alpha
```

```python
scaled_fd001=homogeneous_scaling(fd001_arr)
print(scaled_fd001)
scaled_fd001_df=pd.DataFrame(scaled_fd001,columns=fd001_cols)
scaled_fd001_df.to_csv('homogeneous_scaling_fd001.csv')
```
