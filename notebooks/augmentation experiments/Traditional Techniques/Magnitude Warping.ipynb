Magnitude Warping is Applying a smooth, varying scaling factor along the time axis using a cubic spline

```python
import numpy as np
import pandas as pd
from scipy.interpolate import CubicSpline
```

```python
def magnitude_warping(X, n_knots=4, sigma=0.2):
    """
    X : (seq_len, n_features)
    returns magnitude-warped version
    """
    seq_len = X.shape[0]
    knots = np.linspace(0, seq_len-1, n_knots).astype(int)
    # Generate scaling factors at knots from normal distribution with mean=1
    knot_values = np.random.normal(loc=1.0, scale=sigma, size=n_knots)
    # Cubic spline interpolation
    cs = CubicSpline(knots, knot_values, bc_type='natural')
    alpha_t = cs(np.arange(seq_len))   # shape (seq_len,)
    # Apply to each feature (broadcast)
    return X * alpha_t[:, np.newaxis]
```
